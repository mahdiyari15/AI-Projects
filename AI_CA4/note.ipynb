شبکه عصبی فولی کانکتد چیه؟

نوعی معماری شبکه عصبی است که در آن هر نرون (نورون) در یک لایه، به تمام نرون‌های لایه بعدی وصل است. به همین دلیل بهش می‌گن "کاملاً متصل".

وقتی ورودی یک تصویر باشه قبل از اینکه به مدل لدیم اون رو مسطح میکنیم  مثلا اگه تصویر یک ماتریس 32 در 32 باشه ورودی رو به عنوان یک بردار 1024 تایی تبدیل میکنیم هر مقدار داخل این بردار یک عدد بین ۰ تا ۲۵۵ (یا نرمال‌شده بین ۰ و ۱) هست که شدت روشنایی اون پیکسل رو نشون می‌ده.

ساختار کلی :

```python
# Input Layer:    784 نورون (هر کدوم یک پیکسل)
# Hidden Layer1:  128 نورون (Fully Connected)
# Hidden Layer2:   64 نورون (Fully Connected)
# Output Layer:    10 نورون (برای 10 کلاس مثل لباس، کفش، پیراهن و...)
```

در این ساختار:

هر نورون در یک لایه به همه نورون‌های لایه بعدی متصل است.

بین لایه ها از تابع فعالسازی مثل سیگموند استفاده میشه تا مدل بتونه روابط غیر خطی رو هم یاد بگیره

هدف شبکه: یاد بگیره که با تنظیم وزن‌ها، بتونه کلاس تصویر (مثلاً «کفش ورزشی» یا «تی‌شرت») رو درست پیش‌بینی کنه

چون تعداد پارامترها زیاده، مدل ممکنه Overfit کنه یا برای تصاویر بزرگ، خیلی سنگین و کند بشه.

✅ تابع فعال‌سازی چیه؟
تابع فعال‌سازی (activation function) باعث می‌شه شبکه بتونه روابط غیرخطی یاد بگیره.
اگر فقط ترکیب خطی داشتیم، کل شبکه مثل یه معادله ساده خطی می‌شد که توانایی تشخیص الگوهای پیچیده رو نداشت.

✅ مرحله ۰: ورودی شبکه — تصویر مسطح‌شده
فرض کن تصویر ۲۸×۲۸ داریم.

این تصویر می‌شه یه بردار با ۷۸۴ ویژگی (پیکسل).

این بردار می‌ره به عنوان ورودی به شبکه.



✅ مرحله ۱: لایه اول (Input → Hidden1)
چی میشه اینجا؟
این بردار ۷۸۴تایی می‌ره به لایه اول که مثلاً ۱۲۸ نورون داره.

هر نورون در این لایه یک ترکیب خطی از کل ورودی‌ها می‌سازه. یعنی:
b + x1w1 + x2w2 + ... + x784w784 = zi
برای هر نورون، وزن‌ها فرق می‌کنه. بعدش روی 
zi
ک تابع فعال‌سازی (مثل ReLU) اعمال می‌شه:
ReLU(zi) = ai
یعنی:

نورون اول از ورودی یه ویژگی جدید ساخت.

نورون دوم یه ترکیب دیگه ساخت.

و تا نورون ۱۲۸م...

در واقع این لایه ۱۲۸ ویژگی جدید از اون ۷۸۴ پیکسل یاد می‌گیره.



✅ مرحله ۲: لایه دوم (Hidden1 → Hidden2)
فرض کن لایه دوم ۶۴ نورون داره.

حالا اون ۱۲۸ ویژگی قبلی دوباره می‌رن توی یه لایه که:

برای هر نورونش یه ترکیب جدید می‌سازه.

یعنی الان داریم ترکیبِ ترکیب‌ها رو می‌سازیم.

و با ReLU یا هر تابع دیگه، غیرخطی‌ش می‌کنیم.

این لایه در واقع می‌تونه روابط پیچیده‌تری بین ویژگی‌ها کشف کنه.



🔁 مراحل بعدی (Hidden3 تا Hidden10)
اگه ۱۰ تا لایه مخفی داشته باشی، دقیقاً همین روند تکرار می‌شه:

هر لایه از خروجی لایه قبل ویژگی‌های جدیدتر و انتزاعی‌تر استخراج می‌کنه.

لایه‌های پایین‌تر چیزهایی مثل لبه و رنگ و بافت (در تصویر) یاد می‌گیرن.

لایه‌های بالاتر (نزدیک خروجی) یاد می‌گیرن که مثلاً "این پیکسل‌ها با هم ترکیب بشن احتمالاً یه کفشه!"

✅ مرحله نهایی: لایه خروجی (Output Layer)
فرض کن مسئله ما کلاسه‌بندی بین ۱۰ کلاس هست (مثل Fashion-MNIST):

پس لایه خروجی باید ۱۰ نورون داشته باشه.

هر نورون خروجی احتمال تعلق تصویر به یک کلاس رو نشون می‌ده.

این نورون‌ها با یه تابع به نام Softmax نرمال می‌شن تا خروجی‌شون مجموعش ۱ باشه:

یه تشبیه ساده:
فرض کن می‌خوای از یه عکس بفهمی که "آیا این یه کفشه یا پیراهن؟"

لایه اول: فقط رنگ و روشنی پیکسل‌ها رو نگاه می‌کنه.

لایه دوم: سعی می‌کنه لبه‌ها یا بافت‌ها رو بفهمه.

لایه سوم: می‌گه "اه، این الگو شبیه کفشه!"

لایه آخر: می‌گه "من ۸۵٪ مطمئنم که این یه کفشه."

✔️ بین لایه‌ها یک گراف کامل دو بخشی داریم: هر نورون لایه قبل به همه‌ی نورون‌های لایه بعد وصل میشه.

✅ جمع‌بندی نهایی و نسخه بهبود‌یافته:
در یک شبکه عصبی Fully Connected، ابتدا تصویر ورودی (مثلاً ۳۲×۳۲) به یک بردار ۱۰۲۴تایی مسطح می‌شود. این بردار وارد لایه‌ی ورودی می‌شود، که خودش فقط داده‌ها را نگه می‌دارد (و وزن ندارد). سپس، در هر لایه بعدی (که Fully Connected است)، هر نورون وزن‌هایی برای تمام خروجی‌های لایه‌ی قبل دارد. نورون‌ها مجموع وزن‌دار ورودی‌هایشان را محاسبه می‌کنند، بایاس خود را اضافه می‌کنند، و حاصل را از یک تابع فعال‌سازی عبور می‌دهند. این روند در تمام لایه‌ها ادامه می‌یابد تا به لایه‌ی نهایی برسیم. در لایه‌ی آخر، به اندازه‌ی تعداد کلاس‌ها نورون داریم. خروجی این لایه توسط تابع Softmax نرمال می‌شود تا به هر کلاس یک احتمال اختصاص یابد و در نهایت، کلاس با بیشترین احتمال به عنوان خروجی شبکه در نظر گرفته می‌شود.

شبکه عصبی پیچشی RNN چیه؟

فرض کن ورودی ما یک عکس رنگی ۳۲×۳۲ پیکسل است (۳ کانال رنگی: قرمز، سبز، آبی)
مرحله ۱: لایه پیچشی (Convolutional Layer)
فرض کن یک فیلتر یا کرنل ۳×۳ داریم (یک پنجره کوچک ۳×۳ پیکسل)

این فیلتر روی تصویر حرکت می‌کند از چپ به راست، بالا به پایین (مثل اسکن کردن)

در هر موقعیت، این فیلتر ضرب نقطه‌ای (element-wise multiplication) با آن قسمت از تصویر انجام می‌دهد و همه مقادیر را جمع می‌کند تا یک عدد به دست بیاورد

نتیجه این اعداد یک تصویر جدید به نام نقشه ویژگی (Feature Map) می‌سازد که اندازه‌اش کمی کوچکتر از تصویر اصلی است (مثلاً ۳۰×۳۰، بسته به اندازه فیلتر و تنظیمات)

مثال ساده:

فرض کن فیلتر ۳×۳ رو گذاشتی روی گوشه بالای سمت چپ تصویر (۳×۳×۳ چون ۳ کانال دارد)

هر مقدار پیکسل در این پنجره با وزن فیلتر ضرب می‌شود

این ۳×۳×۳ عدد ضرب شده جمع می‌شوند و عدد نهایی می‌شود یک مقدار برای آن نقطه در نقشه ویژگی

مرحله ۲: تابع فعال‌سازی (مثلاً ReLU)
بعد از محاسبه نقشه ویژگی، یک تابع غیرخطی مثل ReLU روی هر عدد اعمال می‌شود

ReLU هر عدد منفی را صفر می‌کند و عدد مثبت را نگه می‌دارد

این کار باعث می‌شود مدل بتواند الگوهای پیچیده‌تری را یاد بگیرد

مرحله ۳: لایه نمونه‌برداری یا Pooling
حالا فرض کن می‌خواهیم اندازه نقشه ویژگی را کوچک‌تر کنیم تا محاسبات سریع‌تر شود و مدل مقاوم‌تر شود

مثلاً با Max Pooling 2×2: هر ۲×۲ بخش از نقشه ویژگی را نگاه می‌کنیم و بزرگ‌ترین مقدار را نگه می‌داریم

این باعث می‌شود اندازه نقشه نصف شود (مثلاً از ۳۰×۳۰ شود ۱۵×۱۵)

مرحله ۴: تکرار مراحل قبل
چند لایه پیچشی و pooling پشت سر هم قرار می‌گیرد

هر لایه ویژگی‌های بیشتری و سطح بالاتری (جزئیات بیشتر) را یاد می‌گیرد

مرحله ۵: لایه‌های Fully Connected (آخر شبکه)
پس از چند لایه پیچشی و pooling، نقشه ویژگی را "صاف" (flatten) می‌کنیم یعنی همه اعداد را در یک خط در می‌آوریم

این خط اعداد به چند لایه شبکه عصبی معمولی (Fully Connected) داده می‌شود

در نهایت یک خروجی به اندازه تعداد کلاس‌ها (مثلاً ۱۰ کلاس در CIFAR10) می‌گیریم

خروجی نشان‌دهنده احتمال هر کلاس برای تصویر ورودی است

چرا اندازه عکس رو کوچکتر می‌کنیم؟
کم کردن محاسبات: هر چقدر تصویر بزرگ‌تر باشه، مدل باید خیلی محاسبات بیشتری انجام بده. با کوچکتر کردن اندازه، مدل سریع‌تر آموزش می‌بینه.

تمرکز روی ویژگی‌های مهم: با لایه‌های پیچشی و نمونه‌برداری (pooling)، مدل یاد می‌گیره ویژگی‌های مهم‌تر و کلان‌تر تصویر رو حفظ کنه و نویزها یا جزئیات بی‌ربط رو کنار بزاره.

کاهش Overfitting: وقتی مدل خیلی دقیق روی جزییات ریز عکس تمرکز نکنه، بهتر می‌تونه روی داده‌های جدید هم عملکرد خوبی داشته باشه.

اون اعدادی که از لایه‌های پیچشی به دست میاد رو چی کار می‌کنیم؟
این اعداد نقشه ویژگی (Feature Maps) هستند و هر کدومشون یک جنبه از تصویر رو توصیف می‌کنه.

مثلا ممکنه یک فیلتر خاص گوشه‌ها رو تشخیص بده، فیلتر دیگه لبه‌ها رو، یکی بافت‌های خاص رو، و غیره.

وقتی این نقشه‌های ویژگی رو پشت سر هم می‌سازیم، مدل به تدریج یاد می‌گیره که چه الگوهایی در تصویر وجود داره.

استفاده نهایی این اعداد:
این نقشه‌های ویژگی رو بعد از چند لایه صاف می‌کنیم (flatten) یعنی همه این اعداد رو می‌چینیم پشت سر هم به صورت یک بردار بزرگ.

این بردار میره به لایه‌های نهایی که مثل یه شبکه عصبی ساده (Fully Connected) عمل می‌کنن و یاد می‌گیرن بر اساس این ویژگی‌ها به هر کلاس یک امتیاز یا احتمال اختصاص بدن.

در نهایت مدل تصمیم می‌گیره که تصویر مثلا "گربه" هست یا "سگ" یا "ماشین".

یه مثال خیلی ساده
فرض کن می‌خوای یه عکس رو تشخیص بدی

لایه پیچشی میگه: اینجا یه لبه هست، اینجا یه گوشه، اینجا یه رنگ خاص

لایه‌های بعدی ترکیب این اطلاعات رو یاد می‌گیرن: مثلا لبه‌ها کنار هم تشکیل یه چشم رو دادن

آخرش مدل می‌گه: خب، این تصویر خیلی شبیه یه گربه است چون چشم‌ها، گوش‌ها و بافت پشم مشخصه
